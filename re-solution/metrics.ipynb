```python
from sklearn.metrics import (
    explained_variance_score,
    max_error,
    mean_absolute_error,
    mean_absolute_percentage_error,
    r2_score,
    root_mean_squared_error,
    root_mean_squared_log_error,
)
```

```python
rad = [1, 2, 4, 8, 16, 32, 64, 128]
```

```python
bhom = [
    8.77e-25, 4.81e-25, 2.62e-25, 1.60e-25,
    1.04e-25, 6.62e-26, 5.16e-26, 4.38e-26
]
```

```python
def pl_fn(x):
    return 8.43e-25 * x**(-0.926) + 3.46e-26

pl_fit = [pl_fn(x) for x in rad]
```

```python
print(
    max_error(bhom, pl_fit),
    mean_absolute_error(bhom, pl_fit),
    mean_absolute_percentage_error(bhom, pl_fit)
)

print(
    root_mean_squared_error(bhom, pl_fit),
    root_mean_squared_log_error(bhom, pl_fit)
)

print(
    r2_score(bhom, pl_fit),
    explained_variance_score(bhom, pl_fit)
)
```

```python
import matplotlib.pyplot as plt
```

```python
residual = [x - y for x, y in zip (bhom, pl_fit)]
```

```python
plt.scatter(pl_fit, residual)
plt.xscale('log')
```

```python
n = [ 0.8, 1.0, 1.2, 1.4 ]
```

```python
chi_pred = []
for cc in chi:
    tmp = []
    for ni in n:
        tmp.append(cc[1] / ni)
    
    chi_pred.append(tmp)
```

```python
chi_pred
```

```python
for x, y in zip(chi, chi_pred):
    plt.scatter(n, x)
    plt.plot(n, y, marker='')

plt.yscale('log')
plt.show()
```

```python
print(
    mean_absolute_error(chi, chi_pred),
    mean_absolute_percentage_error(chi, chi_pred)
)

print(
    root_mean_squared_error(chi, chi_pred),
    root_mean_squared_log_error(chi, chi_pred)
)

print(
    r2_score(chi, chi_pred),
    explained_variance_score(chi, chi_pred)
)
```
