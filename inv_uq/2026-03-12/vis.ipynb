```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rcupd = {
    #'figure.figsize': (5, 4),
    'text.usetex': True,
    'font.family': 'serif',
    'font.serif': 'cm',
    'font.size': 12,
    'axes.formatter.limits': (-3, 3),
}
plt.rcParams.update(rcupd)
```

```python
jar = pd.read_csv('2025-11-04/RERTR5_V6018G.csv')
```

```python
fig, axes = plt.subplots(nrows=3, ncols=3, sharey=True, figsize=(10, 10))

for i, ax in enumerate(axes.flatten()):
    ax.scatter(jar.iloc[:,i+1], jar.iloc[:,-2],
               color=plt.cm.jet(0.2),
               s=1, rasterized=True)
    ax.set_xlabel(jar.columns[i+1])
    #ax.set_ylabel(jar.columns[-1])

fig.supylabel(r'Fuel Swelling (\%)')
plt.tight_layout()
plt.show()
#plt.savefig('scatter_lo.pdf')
```

```python
plt.figure(figsize=(12,12))
sns.heatmap(jar.corr(), annot=True, cmap='coolwarm')
plt.show()
```
