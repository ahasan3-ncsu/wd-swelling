# Vis

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rcupd = {
    #'figure.figsize': (5, 4),
    'text.usetex': True,
    'font.family': 'serif',
    'font.serif': 'cm',
    'font.size': 12,
    'axes.formatter.limits': (-3, 3),
}
plt.rcParams.update(rcupd)
```

# RERTR-12

```python
jar = pd.read_csv('RERTR12_9_parameters.csv')
```

```python
fig, axes = plt.subplots(nrows=3, ncols=3, sharey=True, figsize=(10, 10))

for i, ax in enumerate(axes.flatten()):
    ax.scatter(jar.iloc[:,i], jar.iloc[:,-3],
               color=plt.cm.jet(0.2),
               s=1, rasterized=True)
    ax.set_xlabel(jar.columns[i+1])
    #ax.set_ylabel(jar.columns[-1])

fig.supylabel(r'Fuel Swelling (\%)')
plt.tight_layout()
plt.show()
#plt.savefig('scatter_lo.pdf')
```

```python
fig, axes = plt.subplots(nrows=3, ncols=3, sharey=True, figsize=(10, 10))

for i, ax in enumerate(axes.flatten()):
    ax.scatter(jar.iloc[:,i], jar.iloc[:,-2],
               color=plt.cm.jet(0.2),
               s=1, rasterized=True)
    ax.set_xlabel(jar.columns[i+1])
    #ax.set_ylabel(jar.columns[-1])

fig.supylabel(r'Porosity')
plt.tight_layout()
plt.show()
#plt.savefig('scatter_lo.pdf')
```

```python
plt.scatter(jar.iloc[:,-3], jar.iloc[:,-2])
plt.xlabel('Fuel swelling')
plt.ylabel('Porosity')
plt.show()
```
